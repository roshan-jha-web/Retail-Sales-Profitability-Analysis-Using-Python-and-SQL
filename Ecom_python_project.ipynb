```python
# I want to consider Not-Availabe, unkown as NA only
df1 = pd.read_csv('orders_data.csv', na_values=['Not Available', 'unknown'])
df1['Ship Mode'].unique()
```

```python
df1.head()
```

```python
# calculating selling price
df1['Selling_Price'] = df1['List Price'] - df1['List Price']*(df1['Discount Percent'] / 100)
df1['Selling_Price']
```

```python
df1.head(5)
```

```python
df1['Profit'] = df1['Selling_Price'] - df1['cost price']
df1['Profit']
```

```python
df1.head()
```

```python
df1.rename(columns={'Profit':'Unit_Profit', 'Selling_Price':'Unit_Selling_Price'}, inplace=True)
```

```python
df1.columns
```

```python
df1.dtypes
```

```python
#changing datatype of order date column from str to datetime
```

```python
df1['Order Date'] = pd.to_datetime(df1['Order Date'], format='%d-%m-%Y')
```

```python
df1.dtypes
```

```python
df1.drop(columns= ['cost price', 'List Price', 'Discount Percent'], inplace=True)
```

```python
#calculating total Profit in each order
```

```python
df1['Total Profit'] = df1['Quantity'] * df1['Unit_Profit']
df1['Total Profit']
```

```python
#Find month over month growth compariosn for 2022 and 2023 
```

```python
a = df1.groupby(['Month','Year'])['Total_Sale'].sum().reset_index()
```

```python
pivot_table = a.pivot_table(index='Month', columns='Year', values='Total_Sale', aggfunc='sum')
pivot_table
```

```python
pivot_table = pivot_table.reset_index()
pivot_table
```

```python
pivot_table['YoY_growth'] = pivot_table[2023]-pivot_table[2022]
pivot_table
```
